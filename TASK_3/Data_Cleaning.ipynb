```python
import pandas as pd
```

```python
data = {
    "Players": [
        "Joe Root",
        "Virat Kohli",
        "Rohit Sharma",
        "Jos Buttler",
        "Virat Kohli"
    ],

    "Runs": [
        "7198",
        "13848",
        "11168",
        None,
        "13848"
    ],

    "Wickets": [
        "68",
        "5",
        None,
        "0",
        "5"
    ],

    "Matches": [
        "100",
        "150",
        "120",
        "80",
        "150"
    ]
}

df = pd.DataFrame(data)

df.to_csv("cricket_dirty.csv", index=False)
print("Orginal Dataset:")
display(df)
```

##Missing values

```python
print("Missing Values in the Dataset:")
print(df.isnull())
```

##Missing values count

```python
print("missing values count in the dataset:")
print(df.isnull().sum())
```

##Remove or fill missing values

```python
df["Runs"]=df["Runs"].fillna("0")
df["Wickets"]=df["Wickets"].fillna("0")
print("Dataset after filling missing values:")
print(df)
```

##Identify duplicate records

```python
print("Duplicated Records in the Dataset:")
print(df.duplicated())
```

##Remove duplicate records

```python
df=df.drop_duplicates()
print("Dataset after removing duplicates:")
print(df)
```

##Correct inappropriate data types

```python
df["Runs"] = df["Runs"].astype(int)
df["Wickets"] = df["Wickets"].astype(int)
df["Matches"] = df["Matches"].astype(int)

print("Corrected Data Types:")
print(df.dtypes)
```

##Display the cleaned dataset

```python
print("Final Cleaned Dataset:")
display(df)
```

```python
df.to_csv("cricket_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
