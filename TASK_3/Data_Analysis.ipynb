```python
import pandas as pd
```

```python
b_df=pd.read_csv("cricket_dirty.csv")
print("dataset Before cleaning:")
print(b_df)
print("Missing Values in the Dataset:")
print(b_df.isnull())
print("missing values count in the dataset:")
print(b_df.isnull().sum())
print("Total Duplicates in the Dataset:")
print(b_df.duplicated().sum())
df=b_df.copy()
df["Runs"]=df["Runs"].fillna("0")
df["Wickets"]=df["Wickets"].fillna("0")
df.drop_duplicates(inplace=True)
df["Runs"]=df["Runs"].astype(int)
df["Wickets"]=df["Wickets"].astype(int) 
df["Matches"]=df["Matches"].astype(int)
df.to_csv("cricket_cleaned.csv", index=False)
print("Dataset After Cleaning:")
print("\nCleaned Dataset:")
display(df)

print("\nMissing Values Count:")
print(df.isnull().sum())

print("\nTotal Duplicate Records:")
print(df.duplicated().sum())

print("\nCorrected Data Types:")
print(df.dtypes)
```
