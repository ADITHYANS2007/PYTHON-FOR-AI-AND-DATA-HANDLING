```python
import pandas as pd
import matplotlib.pyplot as plt
```

1. Bar Chart
The bar chart compares the total runs scored by different cricket players. It helps identify the player with the highest and lowest runs.

```python
data={
    "Players":["Joe root","Virat Kholi","Steve Smith","Kane Williamson","Babar Azam"],
    "Runs":[10000,12000,9000,8000,7000],
    "Wickets":[200,250,150,100,50]
}
df=pd.DataFrame(data)
print(df)
##Bar Chart
plt.figure(figsize=(8, 5))

plt.bar(df["Players"], df["Runs"])

plt.title("Runs Scored by Cricket Players")
plt.xlabel("Players")
plt.ylabel("Runs")

plt.xticks(rotation=30)
plt.savefig("Bar_Chart.png")
plt.show()
```

2. Line Chart
The line chart represents the variation in runs scored by different cricket players. It helps visualize the differences in their performances.

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Players"], df["Runs"], marker='o')

plt.title("Runs Comparison of Cricket Players")
plt.xlabel("Players")
plt.ylabel("Runs")

plt.grid()
plt.savefig("Line_Chart.png")
plt.show()
```

3. Histogram
The histogram shows the distribution of runs scored by the players across different intervals. It helps understand how the runs are spread among the players.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Runs"], bins=4, edgecolor="black")

plt.title("Distribution of Runs")
plt.xlabel("Runs")
plt.ylabel("Number of Players")
plt.savefig("Histogram.png")
plt.show()
```

4. Scatter Plot
The scatter plot shows the relationship between runs scored and wickets taken by each player. Each point represents one cricket player.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Runs"], df["Wickets"])

plt.title("Runs vs Wickets")
plt.xlabel("Runs")
plt.ylabel("Wickets")
plt.savefig("Scatter_Plot.png")
plt.show()
```
