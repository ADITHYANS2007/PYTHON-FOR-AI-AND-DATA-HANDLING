```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Student_Performance.csv")
print(df)
```

```python
df.isnull().sum()
```

```python
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())

df["Science"] = df["Science"].fillna(df["Science"].mean())
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates()
```

```python
df=df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
print(df)
```

```python
df.info()
```

##Total Marks

```python
df["Total_Marks"] = df["Maths"] + df["Science"] + df["English"]

print(df[["Name", "Total_Marks"]])
```

##Average Marks

```python
df["Average_Marks"] = df["Total_Marks"] / 3

print(df[["Name", "Average_Marks"]])
```

##Overall Average

```python
print("Overall Average Marks:", df["Average_Marks"].mean())
```

##highest marks

```python
print("Highest Marks:", df["Average_Marks"].max())
```

##Lowest Marks

```python
print("Lowest Marks:", df["Average_Marks"].min())
```

##students who passed

```python
passed = df[
    (df["Maths"] >= 40) &
    (df["Science"] >= 40) &
    (df["English"] >= 40)
]

print(passed[["Name", "Maths", "Science", "English"]])

print("Number of Students Passed:", len(passed))
```

##students who failed

```python
failed = df[
    (df["Maths"] < 40) |
    (df["Science"] < 40) |
    (df["English"] < 40)
]

print(failed[["Name", "Maths", "Science", "English"]])

print("Number of Students Failed:", len(failed))
```

##OVERALL ANALYSIS

```python
print("Overall Analysis:")

print("Total Students:", len(df))

print("Average Marks:", df["Average_Marks"].mean())

print("Highest Marks:", df["Average_Marks"].max())

print("Lowest Marks:", df["Average_Marks"].min())

print("Students Passed:", len(passed))

print("Students Failed:", len(failed))
```

Bar Chart

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Name"], df["Average_Marks"])

plt.xlabel("Student Name")
plt.ylabel("Average Marks")

plt.title("Student Average Marks")

plt.xticks(rotation=90)
plt.savefig("Bar_chart.png")
plt.show()
```

Pie Chart

```python
labels = ["Passed", "Failed"]

sizes = [len(passed), len(failed)]

plt.figure(figsize=(6, 6))

plt.pie(sizes, labels=labels, autopct="%1.1f%%",
        startangle=90)

plt.title("Student Pass vs Fail Percentage")
plt.savefig("Pie_chart.png")
plt.show()
```

Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Average_Marks"], bins=5, edgecolor="black")

plt.xlabel("Average Marks")
plt.ylabel("Number of Students")

plt.title("Distribution of Student Average Marks")
plt.savefig("Histogram.png")
plt.show()
```
